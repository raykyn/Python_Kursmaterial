# SLSP Abfrage per SRU

Code übernommen und angepasst aus: https://github.com/ub-unibe-ch/ds-pytools/blob/main/SRU/Katalogabfrage_via_SRU-query.ipynb

```python
import requests
from lxml import etree as et
```

```python
base_url = 'https://slsp-ube.alma.exlibrisgroup.com/view/sru/41SLSP_UBE'
```

Mehr Infos zum Bau der Queries siehe CQL: https://www.loc.gov/standards/sru/cql/

Um die Felder einzusehen, verwende z.B. die SRU-Abfrage der SLSP mit `?version=1.2&operation=explain`

```python
query = "alma.creator=Bakunin"
```

```python
query = "alma.title=Bakunin and alma.main_pub_date>=2000-01-01"
```

```python
params = {'version': '1.2',
        'operation': 'searchRetrieve',
        'recordSchema' : 'marcxml',
        'maximumRecords': '50',
        'query': query
        }
```

```python
r = requests.get(base_url, params=params)

print(r)
```

```python
xml = et.fromstring(r.content)

print(xml)
```

# XML durchsuchen

```python
# wir speichern das XML mal ab um es uns in Ruhe anzusehen

et.ElementTree(xml).write('record.xml', pretty_print=True, xml_declaration=True, encoding='UTF-8')
```

```python
ns = {
    'srw': 'http://www.loc.gov/zing/srw/',
    'marc': 'http://www.loc.gov/MARC21/slim'
}
```

```python
records = xml.xpath('./srw:records/srw:record/srw:recordData/marc:record', namespaces=ns)

print(f'Es wurden {len(records)} Datensätze gefunden.')
```

```python
# print out all titles
info = []

for record in records:
    print("="*40)
    titles = record.xpath('./marc:datafield[@tag="245"]/marc:subfield[@code="a"]', namespaces=ns)
    for title in titles:
        print(title.text) 
```
