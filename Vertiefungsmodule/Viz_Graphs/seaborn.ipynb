## Liniendiagramm

Hier seht ihr ein einfaches erstes Beispiel für ein Liniendiagramm (eng: *Lineplot*). Dieser bildet die einzelnen Datenpunkte - in diesem Fall die Höhe des Trinkgelds als x- und die Höhe der Rechnung als y-Koordinate - ab und verbindet die Punkte durch eine durchgängige Linie.

Ist das Dataframe gut formatiert, ist die Visualisierung mit Seaborn sehr einfach. Unter dem *data*-Parameter übergibt man den Datensatz, also das Dataframe. Die Parameter *x* und *y* geben wir die Spaltennamen an, die die Daten auf x- und y-Achse darstellen. In diesem Fall also das Verhältnis zwischen Trinkgeld und totalem Rechnungsbetrag. Viele der Dinge, die ihr im Vertiefungsnotebook zu Pandas schon kennengelernt habt, sind hier je nach Anwendungsfall möglich und nützlich - beispielsweise könnt ihr eure Daten zuerst mit Pandas gruppieren/filtern und sie dann erst visualisieren.

Weitere Infos und Parameter findet ihr in der offiziellen Dokumentation (https://seaborn.pydata.org/generated/seaborn.lineplot.html).

```python
sns.lineplot(data=tips, x="tip", y="total_bill")
```

Der Plot zeigt auf, dass es einen Zusammenhang zwischen der Höhe der Rechnung und der Höhe des Trinkgeldes gibt: Je höher der Betrag, desto höher ist im Schnitt das Trinkgeld - auch wenn es vereinzelt Ausreisser gibt. Das ist bei diesen Daten über Tringkeld nicht überraschend sein. In realen Daten kann es aber durchaus Zusammenhänge geben, die man anfangs gar nicht vermuten würde. 

```python
# Erstelle hier einen lineplot, der die Höhe der Rechnung auf der y- und die Grösse (size) auf der x-Achse zeigt

sns.lineplot()
```

Im folgenden Plot stellen wir die Daten für die beiden Geschlechter unterschiedlich da. Um die Geschlechter zu unterscheiden, verwendet man "hue" und übergibt den Spaltennamen. Dadurch formt Seaborn automatisch Gruppen zu den verschieden Werten, welche in dieser Spalte vorkommen.

Mit dem Parametern "style" kann ähnlich wie die Farbgebung auch die Darstellung der Linie angepasst werden. Mit "markers" hingegen kann aktiviert werden, dass jeweils auf den Schritten der x-Achse Punkte angezeigt werden um das Diagramm besser lesbar zu machen.

```python
sns.lineplot(data=tips, x="size", y="total_bill", hue="sex", style="sex", markers=True)
```

## Histogramm

Histogramme verwendet man, um die Verteilung von Daten darzustellen. In folgendem Beispiel ist dargestellt, wie häufig die verschiedenen Beträge an Trinkgeld gegeben wurden. Hier wurde in 10 Gruppen (sogeannte "bins") unterteilt. Man kann die Unterteilung aber auch spezifischer angeben - z.B. in dem man selbst die einzelnen Start und Endpunkte der Gruppen festlegt. In diesem Fall müssen die Gruppen nicht alle gleich gross sein, so wie sie hier automatisch erstellt worden sind. Weitere Infos und Parameter findet ihr in der offiziellen Dokumentation (https://seaborn.pydata.org/generated/seaborn.histplot.html).

Die darübergelegte Linie - die mit kde=True aktiviert wird - zeigt die mathematische Approximation der zugrunde liegenden Verteilung an.

```python
sns.histplot(data=tips, x="tip", bins=10, kde=True)
```

### Titel setzen mit Mathplotlib

In den obrigen Beispielen wurden die Achsenbeschriftungen automatisch von Seaborn erzeugt. Das funktioniert sogar in der Praxis meistens gut.
Es kann jedoch sein, dass ihr eine andere Beschriftungen bevorzugen würdet, die Schriftgrösse zu klein ist oder ihr dem Plot als ganzen einen Titel geben möchtet. Dies geht nicht direkt über Seaborn, sondern hierzu muss die Matplotlib verwendet werden, die wir in der nächsten Zelle importieren.

```python
import matplotlib.pyplot as plt
```

Im nächsten Beispiel übersetzen wir den letzten Plot auf Deutsch. Dazu setzen wir die Achsentitel und geben dem gesamten Plot eine Überschrift. Bei der Überschrift setzen wir zusätzlich die Schriftgrösse, damit sie genügend gross ist.

```python
sns.histplot(data=tips, x="tip", bins=10, kde=True)
plt.title('Verteilung der Höhe der Trinkgelder', fontsize=20)
plt.xlabel('Höhe des Trinkgelds')
plt.ylabel('Anzahl der Rechnungen')
```

Manchmal sind die automatischen Punkte, an denen Werte der x-Achse zur Orientierung beschriftet sind, unglücklich gewählt. Dies lässt sich auch mit der Matplotlib beheben.

In folgendem Beispiel sind die Daten sehr unregelmässig verteilt und wir wollen ausserdem keine Beschriftung mit Zahlen, sondern nur mit den Bezeichnungen der Messungen. Im automatisch erzeugten Plot sieht dies so aus:

```python
x_positions = [1, 5, 6, 15, 20]
y_values = [10, 25, 15, 30, 40]
labels = ['A', 'B', 'C', 'D', 'E']

plt.plot(x_positions, y_values, marker='o', linestyle='--', color='blue')
```

Jetzt setzen wir die Beschriftungen an der x-Achse - die ticks - neu, damit sie dem entsprechen, wie wir es uns vorstellen. *x_positions* gibt dabei die Positionen an, an denen beschriftet werden soll und labels wird verwendet, um die Beschriftung von Zahlen auf die Benennung der Messungen umzustellen.

```python
x_positions = [1, 5, 6, 15, 20]
y_values = [10, 25, 15, 30, 40]
labels = ['A', 'B', 'C', 'D', 'E']

plt.plot(x_positions, y_values, marker='o', linestyle='--', color='blue')

plt.xticks(
    ticks=x_positions,  # Die Positionen, an denen die Ticks gezeichnet werden sollen
    labels=labels       # Die Beschriftungen, die an diesen Positionen angezeigt werden sollen
)
```
